```python
import timeit
```

```python
def fibonacci(n):
    
    for i in range(2, n + 1):
        fib_list[i] = fib_list[i - 1] + fib_list[i - 2]
    return fib_list[n]

def fibonacci_recursive(n):
    
    if n == 0:
        return 0
    if n == 1:
        return 1
    fib_recur_list[n] = fibonacci_recursive(n - 1) + fibonacci_recursive(n - 2)
    return fib_recur_list[n]
```

```python
N = 20
RUNS = 1000
print(f"Given N = {N}\n{RUNS} runs\n")

fib_list = [0] * (N + 1)
fib_list[0] = 0
fib_list[1] = 1
non_recur_time = timeit.timeit(lambda: fibonacci(N), number=RUNS)
```

```python
print(
    f"Fibonacci non-recursive: {fibonacci(N)}\t"
    f"Time: {non_recur_time:.5f}s\t"
    f"Time: O(n)\tSpace: O(1)"
)

fib_recur_list = [0] * (N + 1)
fib_recur_list[0] = 0
fib_recur_list[1] = 1
recur_time = timeit.timeit(lambda: fibonacci_recursive(N), number=RUNS)

print(
    f"Fibonacci recursive:\t {fibonacci_recursive(N)}\t"
    f"Time: {recur_time:.5f}s\t"
    f"Time: O(2^n)\tSpace: O(n)"
)
```
